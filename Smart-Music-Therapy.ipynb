```python
from IPython.display import display, HTML

# Mood → Playlist
music = {
    "sad": {
        "playlist": "https://youtube.com/playlist?list=PL3-sRm8xAzY-w9GS19pLXMyFRTuJcuUjy&si=sZBgF8IUNbT_-3Z4",
        "reason": "Helps you process emotions and feel understood."
    },
    "happy": {
        "playlist": "https://youtube.com/playlist?list=PLx2Jv96o522ORh69HaDKClrz2Midqj8AE&si=gAE55Ed_hEe1XVov",
        "reason": "Boosts positivity and energy."
    },
    "stressed": {
        "playlist": "https://youtube.com/playlist?list=PLXOoW47yyN2cn0XyBK_WOagu69X-g3iAW&si=anGF5It8ypBDOHAm",
        "reason": "Calms your mind and reduces stress."
    },
    "motivated": {
        "playlist": "https://youtube.com/playlist?list=PLmgutjZvzLyryoakC3VAlDptXy4ChQGC3&si=u3zptp11z2WNxOPe",
        "reason": "Increases motivation and confidence."
    }
}

def detect_mood(text):
    text = text.lower()
    if "sad" in text or "cry" in text:
        return "sad"
    elif "happy" in text or "excited" in text:
        return "happy"
    elif "stress" in text or "tired" in text:
        return "stressed"
    elif "angry" in text or "motivate" in text:
        return "motivated"
    else:
        return None

print("🎧 Welcome to Smart Music Therapy System (Colab Version)")

user_input = input("\nHow are you feeling today? ")

mood = detect_mood(user_input)

if mood in music:
    print("\nDetected Mood:", mood)
    print("Reason:", music[mood]["reason"])
    print("Click below to open your playlist 🎶👇")

    # Opens clickable link in new tab
    display(HTML(f'<a href="{music[mood]["playlist"]}" target="_blank">▶ Open Playlist</a>'))

else:
    print("Sorry, I couldn't understand your mood.")
```
